```python
# 1) Escreva um programa que peça à pessoa usuária para fornecer dois números e exibir o número maior.

primeiro_numero = int(input("Digite o primeiro número: "))
segundo_numero = int(input("Digite o segundo número: "))

if primeiro_numero > segundo_numero:
    print("Primeiro numero {} é maior do que o segundo número {}".format(primeiro_numero,segundo_numero))
else:
    print("Segundo numero {} é maior do que o primeiro número {}".format(segundo_numero,primeiro_numero))
```

```python
# 2) Escreva um programa que solicite o percentual de crescimento de produção de uma empresa e informe se houve um crescimento (porcentagem positiva) ou decrescimento (porcentagem negativa).

percentual_anterior = int("Digite o percentual anterior: ")
percentual_producao = int("Digite o novo percentual de produção: ")


if percentual_anterior < percentual_producao:
    print("Houve um aumento de percentual positivo em relação ao anterior. Anterior: {} e Novo: {}".format(percentual_anterior,percentual_producao))
else:
    print("Houve um decrescimento de percentual em relação ao anterior. Anterior: {} e Novo: {}".format(percentual_anterior,percentual_producao))
```

```python
# 3) Escreva um programa que determine se uma letra fornecida pela pessoa usuária é uma vogal ou consoante.

letra_usuario = input("Digite a letra:")

letra_tratamento = letra_usuario.lower()

if letra_tratamento in ['a','e','i','o','u']:
    print("A letra usuário: {}, é vogal.".format(letra_tratamento))
```

```python
# 4) Escreva um programa que leia valores médios de preços de um modelo de carro por 3 anos consecutivos e exiba o valor mais alto e mais baixo entre esses três anos.


media_valor1 = float(input("Digite o primeiro valor médio: "))
media_valor2 = float(input("Digite o segundo valor médio: "))
media_valor3 = float(input("Digite o terceiro valor médio: "))

# MAIOR MÉDIA

if media_valor1 > media_valor2 and media_valor1 > media_valor3:
    maior = media_valor1
elif media_valor2 > media_valor1 and media_valor2 > media_valor3:
    maior = media_valor2
else:
    maior = media_valor3

# MENOR MÉDIA

if media_valor1 < media_valor2 and media_valor1 < media_valor3:
    menor = media_valor1
elif media_valor2 < media_valor1 and media_valor2 < media_valor3:
    menor = media_valor2
else:
    menor = media_valor3


print("O valor mais alto entre os 3 anos foi: {}".format(maior))
print("O valor mais baixo entre os 3 anos foi: {}".format(menor))
```

```python
# 5) Escreva um programa que pergunte sobre o preço de três produtos e indique qual é o produto mais barato para comprar.


nome_primeiro_produto = input("Digite o nome do produto 1: ")
primeiro_produto = float(input("Digite o preço do produto 1: "))

segundo_produto = float(input("Digite o preço do produto 2: "))
nome_segundo_produto = input("Digite o nome do produto 2: ")

nome_terceiro_produto = input("Digite o nome do produto 3: ")
terceiro_produto = float(input("Digite o preço do produto 3: "))


if primeiro_produto < segundo_produto and primeiro_produto < terceiro_produto:
    menor_preco = primeiro_produto
    print("O produto {} possui o menor preço: R$ {}".format(nome_primeiro_produto,menor_preco))

elif segundo_produto < primeiro_produto and segundo_produto < terceiro_produto:
    menor_preco = segundo_produto
    print("O produto {} possui o menor preço: R$ {}".format(nome_segundo_produto,menor_preco))

else:
    menor_preco = terceiro_produto
    print("O produto {} possui o menor preço: R$ {}".format(nome_terceiro_produto,menor_preco))
    
```

```python
# 6) Escreva um programa que leia três números e os exiba em ordem decrescente.

primeiro_numero = int("digite o primeiro número: ")
segundo_numero = int("digite o segundo número: ")
terceiro_numero = int("digite o terceiro número: ")

lista_numero = [primeiro_numero,segundo_numero,terceiro_numero]

lista_numero.sort(reverve=True)

for numero in lista_numero:
    print(numero)
```

```python
# 7) Escreva um programa que pergunte em qual turno a pessoa usuária estuda ("manhã", "tarde" ou "noite") e exiba a mensagem "Bom Dia!", "Boa Tarde!", "Boa Noite!", ou "Valor Inválido!", conforme o caso.

turno = input("digite o periodo que você estuda: ")

turno.lower()

if turno == 'manha':
    print("Bom dia! ")
elif turno == 'tarde':
    print("Boa tarde! ")
elif turno == 'noite':
    print("Boa noite! ")
else:
    print("Valor inválido! ")
```

```python
# 8) Escreva um programa que peça um número inteiro à pessoa usuária e determine se ele é par ou ímpar. Dica: Você pode utilizar o operador módulo %.

numero_inteiro = int(input("Digite um número: "))


if numero_inteiro % 2 == 0:
    print("O {} é par! ".format(numero_inteiro))
else:
    print("O {} é impar! ".format(numero_inteiro))
```

```python
# 9) Escreva um programa que peça um número à pessoa usuária e informe se ele é inteiro ou decimal.

numero = float(input("Digite o seu número: "))

if numero == int(numero):
    print("O seu número {} é inteiro".format(numero))
else:
    print("O seu número {} é decimal".format(numero))
```
