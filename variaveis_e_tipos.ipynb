# CALCULADORA COM OPERADORES

```python
# 1. Crie um programa que solicite dois valores numéricos à pessoa usuária e imprima a soma dos dois valores.


primeiro_numero = int(input("Digite o primeiro número: "))
segundo_numero = int(input("Digite o segundo número: "))
soma_numeros = primeiro_numero + segundo_numero

print("A soma de {} e {} é igual a : {}".format(primeiro_numero,segundo_numero,soma_numeros))
```

```python
# 2. Crie um programa que solicite três valores numéricos à pessoa usuária e imprima a soma dos três valores.


primeiro_numero = int(input("Digite o primeiro número: "))
segundo_numero = int(input("Digite o segundo número: "))
terceiro_numero = int(input("Digite o segundo número: "))
soma_numeros = primeiro_numero + segundo_numero + terceiro_numero

print("A soma de {} + {} + {} é igual a : {}".format(primeiro_numero,segundo_numero,terceiro_numero,soma_numeros))
```

```python
# 3. Crie um programa que solicite dois valores numéricos à pessoa usuária e imprima a subtração do primeiro pelo o segundo valor.

primeiro_numero = int(input("Digite o primeiro número: "))
segundo_numero = int(input("Digite o segundo número: "))
sub_numeros = primeiro_numero - segundo_numero 

print("A subtração de {} - {} é igual a : {}".format(primeiro_numero,segundo_numero,sub_numeros))
```

```python
# 4. Crie um programa que solicite dois valores numéricos à pessoa usuária e imprima a multiplicação dos dois valores.


primeiro_numero = int(input("Digite o primeiro número: "))
segundo_numero = int(input("Digite o segundo número: "))
multi_numeros = primeiro_numero * segundo_numero 

print("A multiplicação de {} * {} é igual a : {}".format(primeiro_numero,segundo_numero,multi_numeros))
```

```python
# 5. Crie um programa que solicite dois valores numéricos, um numerador e um denominador, e realize a divisão entre os dois valores. Deixe claro que o valor do denominador não pode ser 0.



primeiro_numero = int(input("Digite o primeiro número: "))
segundo_numero = int(input("Digite o segundo número: "))
divisao_numeros = primeiro_numero / segundo_numero 

print("A divisão de {} / {} é igual a : {}".format(primeiro_numero,segundo_numero,divisao_numeros))
```

```python
# 5. Crie um programa que solicite dois valores numéricos, um operador e uma potência, e realize a exponenciação entre esses dois valores.


primeiro_numero = int(input("Digite o primeiro número: "))
segundo_numero = int(input("Digite o segundo número: "))
expo_numeros = primeiro_numero ** segundo_numero 

print("A exponenciacao de {} ** {} é igual a : {}".format(primeiro_numero,segundo_numero,expo_numeros))
```

```python
# 6. Crie um programa que solicite dois valores numéricos, um numerador e um denominador e realize a divisão inteira entre os dois valores. Deixe claro que o valor do denominador não pode ser 0.


primeiro_numero = int(input("Digite o primeiro número: "))
segundo_numero = int(input("Digite o segundo número: "))
divi_inteira_numeros = primeiro_numero // segundo_numero 

print("A divisao inteira de {} // {} é igual a : {}".format(primeiro_numero,segundo_numero,divi_inteira_numeros))
```

```python
# 7. Crie um programa que solicite dois valores numéricos, um numerador e um denominador, e retorne o resto da divisão entre os dois valores. Deixe claro que o valor do denominador não pode ser 0.

primeiro_numero = int(input("Digite o primeiro número: "))
segundo_numero = int(input("Digite o segundo número: "))
divi_resto_numeros = primeiro_numero % segundo_numero 

print("O resto da divisao de {} % {} é igual a : {}".format(primeiro_numero,segundo_numero,divi_resto_numeros))
```

```python
# 8. Crie um código que solicita 3 notas de um estudante e imprima a média das notas.



primeira_nota = int(input("Digite a primeira nota: "))
segunda_nota = int(input("Digite a terceira nota: "))
terceira_nota = int(input("Digite a terceira nota: "))
media_notas = (primeira_nota + segunda_nota + terceira_nota) / 3 

print("As notas foram : {}, {} e {}. E a média : {}".format(primeira_nota,segunda_nota,terceira_nota,media_notas))
```

```python
# 9. Crie um código que calcule e imprima a média ponderada dos números 5, 12, 20 e 15 com pesos respectivamente iguais a 1, 2, 3 e 4.


numero1 = 5
numero2 = 12
numero3 = 20
numero4 = 15

peso1 = 1
peso2 = 2
peso3 = 3
peso4 = 4

soma_ponderada = (numero1 * peso1) + (numero2 * peso2) + (numero3 * peso3) + (numero4 * peso4)
soma_pesos = peso1 + peso2 + peso3 + peso4

media_ponderada = soma_ponderada / soma_pesos

print("A média ponderada é igual a {}".format(media_ponderada))
```
